# Join

```python
billionaires = Table.read_table('billionaires.csv')
billionaires.show(10)

countries = Table.read_table('countries.csv')
countries.show(10)
```

### What can we join these two tables on?

```python
billionaires.join('citizenship', countries, 'Country')
```

```python
billionaires.join('citizenship', countries, 'Continent')
```

```python
countries.join('Country', billionaires, 'citizenship')
```

```python
b_and_c = billionaires.join('citizenship', countries, 'Country')
```

```python
b_and_c.group('Continent')
```

##### Question: How many billionaries are in a specific industry and Continent?

_Hint: Is there a way we can format the data to see this easily?_

```python
b_and_c.labels
```

```python
b_and_c.group(['Continent', 'industry']).barh('industry', 'count')
```

```python
b_and_c.group(['Continent', 'industry'])
```

```python
b_and_c.pivot('Continent', 'industry')
```

##### Question: What is the average net worth of the billionaries in each specific industry and Continent?


```python
b_and_c.pivot('Continent', 'industry', 'net worth', np.average)
```

##### Question: What is the range of net worths of the billionaries in each specific industry and Continent?
_Hint: It may be helpful to define a function that will compute the range_


```python
b_and_c.where('industry', 'Finance').where('citizenship', 'Brazil').column('net worth')
```

```python
max(b_and_c.where('industry', 'Finance').where('citizenship', 'Brazil').column('net worth'))
```

```python
min(b_and_c.where('industry', 'Finance').where('citizenship', 'Brazil').column('net worth'))
```

```python
def range(x):
    "Given as input an array of numerical values, compute the range"
    return max(x) - min(x)
```

```python
range(b_and_c.where('industry', 'Finance').where('Continent', 'North America').column('net worth'))
```

```python
b_and_c.group(['Continent', 'industry'])
```

```python
max(b_and_c.where('Continent', 'Africa').where('industry', 'Diversified').column('net worth')) - min(b_and_c.where('Continent', 'Africa').where('industry', 'Diversified').column('net worth'))
```
