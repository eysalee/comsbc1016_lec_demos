### Create functions

```python
def age(year):
    "Takes as input the birth year and computes their age as of 2026"
    return 2026 - year
```

```python
def age_as_of(birth_year, current_year=2026):
    "Takes in a two years and computes the age as of the second year"
    return current_year - birth_year
```

### Know when to use grouping vs. pivoting!

```python
# https://corgis-edu.github.io/corgis/csv/skyscrapers/
skyscrapers = Table.read_table('skyscrapers.csv')
skyscrapers.show(6)
```

How can I find the tallest building per material for each city?

```python
skyscrapers.select('city', 'material', 'height').group(['city', 'material'], max)
```

```python
skyscrapers.pivot('material','city',  'height', max)
```
