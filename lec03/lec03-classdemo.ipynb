```python
# This is technically datascience.Table

cones = Table.read_table('cones.csv')
cones
```

```python
cones.show(3)
```

```python
# How to manually create a table

streets = make_array('Broadway', 'Amsterdam', 'Columbus', 'Manhattan')
streets
```

```python
eastside = Table().with_column('Streets', streets)
eastside
```

```python
eastside = eastside.with_column('Blocks from campus', np.arange(4))
```

```python
eastside2 = Table().with_columns('Streets', streets, 'Blocks from campus', np.arange(4))
eastside2
```

```python
# Only give me rows where a condition is met.
# The condition is going to be specified as I give you a column
# And then you take out all the rows in which some predicate is not met
cones.where('Flavor', 'chocolate')
```

```python
# Only give me rows where a condition is met.
# The condition is going to be specified as I give you a column
# And then you take out all the rows in which some predicate is not met
cones.where('Flavor', 'chocolate').drop('Flavor')
```
