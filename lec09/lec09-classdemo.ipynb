```python
def sign(x):
    """determines the sign (+/-) of a numeric input. 
    It should return 'Positive' if a value is positive, 
    'Negative' if it's negative, 
    and 'Neither positive or negative' if it's none of the above."""
    if type(x) != int and type(x) != float:
        # This only runs if it's not an integer and it's not a float
        # This will exit the function
        return "This is not a number"

    # We are guaranteed from the previous conditional that we have a number
    if x > 0:
        return 'Positive'
    elif x < 0:
        return 'Negative'
    else:
        return 'Neither positive nor negative'
```

```python
def is_even(x):
    if x%2 == 0:
        return "even"
    else:
        return "odd"
```

### Control your computations with a `for` loop - examples

```python
np.arange(5)
```

```python
for i in np.arange(10):
    print(str(i) + " is " + is_even(i))
```

```python
running_sum = 0

for i in np.arange(5):
    if is_even(i)=='even':
        running_sum = running_sum + i

print(running_sum)
```
